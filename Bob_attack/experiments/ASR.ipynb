# ASR

```python
import numpy as np
import glob
```

```python
def reconstruction_accuracy_from_distances(distances_path, threshold):
    distances = np.load(distances_path)
    accuracy = np.mean(distances <= threshold)
    # print('Accuracy: ', accuracy)
    return accuracy
```

```python
distance_files = glob.glob('results/*/*/*/*.npy')
distance_files.sort()
print('Distance files: ', distance_files)
```

```python
for distance_file in distance_files:
    threshold = 1.1027  # facenet euclidean_l2
    accuracy = reconstruction_accuracy_from_distances(distance_file, threshold)
    print(f'{accuracy*100:.2f}% {threshold} {distance_file}')
```
